```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import pandas as pd


#abrir el csv /Users/maria/Documents/Business Intehlligence/datasets/AtencionesUrgencia2024.csv para explorar los datos y hacer un análisis de los mismos.
with open('/Users/maria/Documents/Business Intehlligence/datasets/AtencionesUrgencia2024.csv', 'r') as file:
    # Leer el archivo CSV en un DataFrame de pandas
    file.close()
    df = pd.read_csv(file.name, sep=';', encoding='cp1252')

df.head()  # Mostrar las primeras filas del DataFrame para explorar los datos

```

```python
#limpieza de valores nulos 
df.isnull().sum()  # Contar los valores nulos en cada columna
df_limpio = df.dropna(inplace=True)  # Eliminar filas con valores nulos
```

```python
#numero total de filas y columnas
# Crear el DataFrame limpio y mostrar sus dimensiones
df_limpio = df.dropna()
df_limpio.shape
```

```python
#todas las columnas del dataset
df_limpio.columns  # Mostrar todas las columnas del DataFrame
```

```python
#Menores_1	De_1_a_4	De_5_a_14	De_15_a_64	De_65_y_mas

#sumar todas las filas por cada columna para obtener el total de personas atendidas en urgencias por rango de edad.
total_personas = df_limpio[['Menores_1', 'De_1_a_4', 'De_5_a_14', 'De_15_a_64', 'De_65_y_mas']].sum()
print(total_personas)  # Mostrar el total de personas atendidas en urgencias por rango de edad
```

```python
#diagnosticos más frecuentes
#hay una columna de idCausa que contiene el codigo del diagnostio, y en la fila del diagnostico se encuentra la cantidad de personas atendidas por rango etario y un total de personas con ese diagnostico.

diagnosticos_frecuentes = (
    df_limpio
    .groupby(['IdCausa', 'GlosaCausa'], as_index=False)['Total_Personas_Atendidas']
    .sum()
    .sort_values('Total_Personas_Atendidas', ascending=False)
)

diagnosticos_frecuentes
print(diagnosticos_frecuentes)  # Mostrar los diagnosticos más frecuentes y la cantidad de personas atendidas por cada uno
```

```python
#cantidad de IdCausa únicos y sus nombres de diagnosticos asociados
cantidad_id_causa_unicos = df_limpio['IdCausa'].nunique()
print(f'Cantidad de IdCausa únicos: {cantidad_id_causa_unicos}')

# Obtener los nombres de los diagnosticos asociados a cada IdCausa único
nombres_diagnosticos = df_limpio[['IdCausa', 'GlosaCausa']].drop_duplicates()
print('\nNombres de diagnosticos asociados:')
print(nombres_diagnosticos)
```
